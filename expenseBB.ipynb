```python
import pandas as pd
from faker import Faker
import random


fake = Faker()


categories = ['Food', 'Transport', 'Utilities', 'Entertainment', 'Healthcare']
payment_modes = ['UPI', 'Credit Card', 'Debit Card', 'Online Transfer' , 'Cash']
```

```python
def gen_exp_data(num_entries=100):
    data = []
    for _ in range(num_entries):
        expense = {
            "Date": fake.date_this_month(),
            "Category": random.choice(categories),
            "Payment Mode": random.choice(payment_modes),
            "Description": fake.sentence(),
            "Amount": round(random.uniform(10, 1000), 2),
            "Cash back": round(random.uniform(1,50),2)
        }
        data.append(expense)
    
    return pd.DataFrame(data)
```

```python
exp_data = gen_exp_data(100)
exp_data
```

```python
# Create a database and a table to store the expense data
try:
    cursor = connection.cursor()
    
    # Create a database
    cursor.execute("CREATE DATABASE IF NOT EXISTS expensedb1")
    print("Database 'expensedb1' created successfully.")

    # Use the created database
    cursor.execute("USE expensedb1")

    # Create a table
    create_table_query = """
    CREATE TABLE IF NOT EXISTS expenses (
        id INT AUTO_INCREMENT PRIMARY KEY,
        Date DATE,
        Category VARCHAR(50),
        Payment_Mode VARCHAR(50),
        Description TEXT,
        Amount DECIMAL(10, 2),
        Cash_back DECIMAL(10, 2)
        
    )
    """
    cursor.execute(create_table_query)
    print("Table 'expenses' created successfully.")

except Error as e:
    print("Error while creating database or table", e)
```

```python
# Insert the generated fake data into the table
try:
    # Insert data into the expenses table
    for index, row in exp_data.iterrows():
        insert_query = """
        INSERT INTO expenses (Date, Category, Payment_Mode, Description, Amount, Cash_back)
        VALUES (%s, %s, %s, %s, %s, %s)
        """
        cursor.execute(insert_query, (row['Date'], row['Category'], row['Payment Mode'], row['Description'], row['Amount'], row['Cash back']))

    # Commit the transaction
    connection.commit()
    print(f"{cursor.rowcount} records inserted successfully into 'expenses' table.")

except Error as e:
    print("Error while inserting data into MySQL table", e)
```
